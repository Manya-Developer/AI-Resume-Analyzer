# TF-IDF Based Resume–Job Matching

This notebook builds the first large-scale baseline model
for matching resumes with job descriptions using TF-IDF
and cosine similarity

```python
import pandas as pd
import numpy as np
```

```python
train_df = pd.read_csv("../data/processed/train_cleaned.csv")
test_df = pd.read_csv("../data/processed/test_cleaned.csv")
```

```python
print("Training shape:", train_df.shape)
print("Testing shape:", test_df.shape)
```

```python
train_df[["resume_clean", "jd_clean", "label"]].head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
vectorizer = TfidfVectorizer(
    stop_words="english",
    max_features=20000
)
```

```python
all_text = pd.concat([
    train_df["resume_clean"],
    train_df["jd_clean"]
])
```

```python

vectorizer.fit(all_text)
```

```python
resume_vectors = vectorizer.transform(train_df["resume_clean"])

jd_vectors = vectorizer.transform(train_df["jd_clean"])
```

```python
print("Resume vector shape:", resume_vectors.shape)
print("JD vector shape:", jd_vectors.shape)
```
